# 簡介

串列(list)是一個可以儲存多個元素(整數，福點數，或字串)的容器。

串列中每個元素都是使用逗點隔開，左右有中括號[]。

```python
scores=[78,29,56,89,99]
print(scores)
```

## 索引值

我們可以使用索引值(index)來存取串列中的元素。
注意:
索引值由0開始。

```python
print(scores[0])
print(scores[1])
```

## 串列內容的個數

len(a)取得a致個處列元素的個數。
len是length(長度)的意思。

```python
print(len(scores))
print(scores[4])
print(scores[len(scores)-1])
```
