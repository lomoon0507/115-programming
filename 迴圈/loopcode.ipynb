# 九九乘法表

1\*1=1,1\*2=2,1\*3=3,...
2\*1=2,2\*2=4,2\*3=6,...
3\*1=3,3\*2=6,3\*3=9,...

```python
i=1
for j in range(1,6):
    print(f"i*{j}={i*j}",end=",")
print()
i=2
for j in range(1,6):
    print(f"i*{j}={i*j}",end=",")
print()
i=3
for j in range(1,6):
    print(f"i*{j}={i*j}",end=",")
print()

```
