# 九九乘法表

1\*1=1,1\*2=2,1\*3=3,...
2\*1=2,2\*2=4,2\*3=6,...
3\*1=3,3\*2=6,3\*3=9,...

```python
i=1
for j in range(1,6):
    print(f"i*{j}={i*j}",end=",")
print()
i=2
for j in range(1,6):
    print(f"i*{j}={i*j}",end=",")
print()
i=3
for j in range(1,6):
    print(f"i*{j}={i*j}",end=",")
print()

```

# 練習:印出質數

質數:一個數字$n$，除了自己本身，以及1以外，沒有任何數可以除盡他，n就是質數。

```python
n=273
i=2

if 273%i==0:
    print(f"{273}不是質數")
i=3
if 273%i==0:
    print(f"{273}不是質數")
i=4
if 273%i==0:
    print(f"{273}不是質數")
```

```python
for i in range(2,273):
    if 273%i==0:
        print(f"{273}不是質數,因為它可以被{i}除盡")
        break#只要數字能被除盡，就能跳離迴圈
    else:
        print(f"")
```

```python
n=17
isPrime=True
for i in range(2,n):
    if n%i==0:
        print(f"{n}不是質數,因為它可以被{i}除盡")
        isPrime=False
        break#只要數字能被除盡，就能跳離迴圈
if isPrime==True: #判斷我們得假設是否還成立
        print(f"{n}是質數")
```
