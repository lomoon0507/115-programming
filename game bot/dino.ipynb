# 介紹
使用pyautogui來完成小恐龍的遊戲機器人。
小恐龍是在chrome中使用'chrome://dino'啟動。

```python
import pyautogui as pyg
pyg.sleep(2)


#無窮迴圈
while True:
    im=pyg.screenshot()
    obs=False
    for i in range(15):    
        c=im.getpixel((191+i,392))
        if c==(83,83,83):
            obs=True
            break
    if c==(83,83,83):
        pyg.press("space") #啟動
```
