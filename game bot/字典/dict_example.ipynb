# 字典，在python鐘用於查表。

字典的前後是用大括號{}包住
裡面每個元素用逗點(,)隔開。
每個元素都是key:value

```python
students={"peter":175,"Mary":168,"keven":165}
#查學生表格，使用關鍵字(key)keven來查
print(students["keven"])
```

```python
students={
     "peter":[175,70,"高雄"],
     "Mary":[168,50,"嘉義"],
     "keven":[165,55,"高雄"]
}

print(students["Mary"])
print(students["Mary"][0])#取得身高
print(students["Mary"][2])#取得戶籍
```

```python
print("Peter"in students)
print("peter"in students)
print("JoJo"in students)
```

```python
#新增jhon進入students中
students["john"]=[180,99,"台北"]
print(students)
```

```python
#由students中移除keven
del students["keven"]
print(students)
```

```python
#len可以得到字典中元素的個數
print(len(students))
```
